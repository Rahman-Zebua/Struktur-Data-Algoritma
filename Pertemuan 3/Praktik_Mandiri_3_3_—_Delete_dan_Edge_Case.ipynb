```python
#3.Mencoba list dengan lima node. Hapus head, node tengah, dan node terakhir secara berurutan
#Penjelasan outputnya:
#Awalnya: [10, 20, 30, 40, 50]
#Hapus head 10 → [20, 30, 40, 50]
#Hapus node tengah 30 → [20, 40, 50]
#Hapus node terakhir 50 → [20, 40]

data = [10, 20, 30, 40, 50]
del data[0]
print(data)
del data[1]
print(data)
del data[-1]
print(data)
```

```python
#4.Mencoba menghapus nilai yang tidak adadan memastikan program tidak error
data = [10, 20, 30, 40, 50]
nilai = 60
if nilai in data:
    data.remove(nilai)
else:
    print("Nilai tidak ditemukan")
print(data)
```

```python
#5.Setelah setiap operasi, mencoba menampilkan isi list dan jumlah node
#Penjelasan maksud dari output:
#Pada awalnya terdapat 5 node yaitu [10, 20, 30, 40, 50]. Setelah node pertama 10 dihapus, tersisa 4 node. Kemudian node 30 dihapus, sehingga tersisa 3 node. Terakhir, node 50 dihapus sehingga tersisa 2 node. Jadi, setiap kali satu node dihapus, jumlah node dalam list berkurang satu.

data = [10, 20, 30, 40, 50]
del data[0]
print("List:", data)
print("Jumlah node:", len(data))
del data[1]
print("List:", data)
print("Jumlah node:", len(data))
del data[-1]
print("List:", data)
print("Jumlah node:", len(data))
```

```python
#6.Penjelaskan peran previous saat menghapus node yang bukan head
Penjelasan:
Previous berperan untuk menyimpan node yang berada tepat sebelum node yang akan dihapus. Saat menghapus node yang bukan head, previous digunakan untuk menghubungkan node sebelumnya dengan node setelah node yang dihapus. Dengan begitu, node yang ingin dihapus dapat dilewati dari linked list tanpa mengubah hubungan node lainnya.
```

```python
#7.Mencoba melakukan pengujian sendiri untuk menemukan kemungkinan kesalahan implementasi
data = [10, 20, 30, 40, 50]
nilai = 60
if nilai in data:
    data.remove(nilai)
else:
    print("Nilai tidak ditemukan")
print("List:", data)
print("Jumlah node:", len(data))

#Outpu
#Nilai tidak ditemukan
#List: [10, 20, 30, 40, 50]
#Jumlah node: 5

#Penjelasan Output
#Pengujian tambahan dilakukan dengan mencoba menghapus nilai 60 yang tidak terdapat dalam list. Hasilnya program tidak mengalami error dan menampilkan pesan "Nilai tidak ditemukan". Isi list tetap sama dan jumlah node tetap 5. Pengujian ini menunjukkan bahwa program dapat menangani kondisi ketika nilai yang ingin dihapus tidak ada.
```
