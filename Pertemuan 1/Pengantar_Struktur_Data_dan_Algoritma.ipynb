```python
#1.4 Fungsi pencarian linear sederhana
def cari(data, target):
  for i in range(len(data)):
    if data[i] == target:
      return i
  return -1
data = [12, 20, 7, 30]
print(cari(data, 7))
print(cari(data, 99))
```

```python
#1.5 Bentuk awal node yang akan digunakan pada linked list
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None
node1 = Node(10)
print(node1.data)
print(node1.next)
```

```python
#1.7 Pencarian NIM menggunakan list
mahasiswa = [
  ('22001', 'Alya'),
  ('22002', 'Bima'),
  ('22003', 'Citra')
]
def cari_nim(data, nim):
  for item in data:
    if item[0] == nim:
      return item
  return None
print(cari_nim(mahasiswa, '22003'))
```

```python
#1.8 Pencarian NIM menggunakan dictionary
mahasiswa = {
  '22001': 'Alya',
  '22002': 'Bima',
  '22003': 'Citra'
}
print(mahasiswa.get('22003'))
```
