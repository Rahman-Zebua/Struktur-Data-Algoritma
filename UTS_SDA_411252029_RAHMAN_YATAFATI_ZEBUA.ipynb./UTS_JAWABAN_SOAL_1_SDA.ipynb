```python
#3. Buat fungsi merge(left, right) dan merge_sort_nim(data) secara manual untuk mengurutkan data berdasarkan kode mahasiswa secara ascending. Merge Sort wajib menggunakan rekursi.
def merge(left, right):  # Membuat fungsi untuk menggabungkan dua bagian data terurut
    hasil = []  # Menyediakan list kosong untuk menampung hasil penggabungan
    i = 0  # Menentukan indeks awal bagian kiri
    j = 0  # Menentukan indeks awal bagian kanan

    while i < len(left) and j < len(right):  # Membandingkan selama kedua bagian masih memiliki data
        if left[i][0] <= right[j][0]:  # Membandingkan kode mahasiswa pada bagian kiri dan kanan
            hasil.append(left[i])  # Memasukkan data kiri jika kodenya lebih kecil atau sama
            i += 1  # Berpindah ke data kiri berikutnya
        else:  # Dijalankan jika kode mahasiswa bagian kanan lebih kecil
            hasil.append(right[j])  # Memasukkan data kanan ke hasil
            j += 1  # Berpindah ke data kanan berikutnya

    while i < len(left):  # Memeriksa apakah masih ada data tersisa di bagian kiri
        hasil.append(left[i])  # Memasukkan sisa data kiri ke hasil
        i += 1  # Berpindah ke data kiri berikutnya

    while j < len(right):  # Memeriksa apakah masih ada data tersisa di bagian kanan
        hasil.append(right[j])  # Memasukkan sisa data kanan ke hasil
        j += 1  # Berpindah ke data kanan berikutnya

    return hasil  # Mengembalikan data yang sudah digabungkan secara terurut


def merge_sort_nim(data):  # Membuat fungsi Merge Sort untuk mengurutkan data mahasiswa
    if len(data) <= 1:  # Memeriksa kondisi dasar jika data berjumlah nol atau satu
        return data  # Mengembalikan data karena sudah terurut

    tengah = len(data) // 2  # Menentukan posisi tengah untuk membagi data
    left = merge_sort_nim(data[:tengah])  # Mengurutkan bagian kiri dengan pemanggilan rekursif
    right = merge_sort_nim(data[tengah:])  # Mengurutkan bagian kanan dengan pemanggilan rekursif

    return merge(left, right)  # Menggabungkan bagian kiri dan kanan yang sudah terurut


d = 9  # Menentukan digit terakhir NIM

data = [  # Menyimpan data mahasiswa sebelum diurutkan
    ['TI001', 'Alya', 78],  # Data mahasiswa Alya
    ['TI002', 'Bima', 89],  # Data mahasiswa Bima
    ['TI003', 'Citra', 67],  # Data mahasiswa Citra
    ['TI004', 'Danu', 92],  # Data mahasiswa Danu
    ['TI005', 'Eka', 74],  # Data mahasiswa Eka
    ['TI006', 'Fajar', 88],  # Data mahasiswa Fajar
    ['TI007', 'Gina', 81],  # Data mahasiswa Gina
    ['TI008', 'Hadi', 74],  # Data mahasiswa Hadi
    ['TI009', 'Intan', 91],  # Data mahasiswa Intan
    ['TI010', 'Joko', 76],  # Data mahasiswa Joko
    ['TI011', 'Karin', 84],  # Data mahasiswa Karin
    ['TI012', 'Lukman', 79]  # Data mahasiswa Lukman
]

data_terurut = merge_sort_nim(data)  # Mengurutkan data menggunakan Merge Sort manual

print("Data mahasiswa setelah diurutkan:")  # Menampilkan judul hasil pengurutan
for mahasiswa in data_terurut:  # Mengakses setiap data mahasiswa yang sudah terurut
    print(mahasiswa)  # Menampilkan data mahasiswa
```

```python
#5. Buat fungsi binary_search_nim(data_urut, target) secara manual. Fungsi harus mengembalikan indeks hasil dan jumlah perbandingan.
def binary_search_nim(data_urut, target):  # Membuat fungsi Binary Search untuk mencari kode mahasiswa
    kiri = 0  # Menentukan batas indeks paling kiri
    kanan = len(data_urut) - 1  # Menentukan batas indeks paling kanan
    jumlah_perbandingan = 0  # Menginisialisasi jumlah perbandingan

    while kiri <= kanan:  # Mengulang selama batas pencarian masih valid
        tengah = (kiri + kanan) // 2  # Menghitung indeks tengah
        jumlah_perbandingan += 1  # Menambah jumlah perbandingan

        if data_urut[tengah][0] == target:  # Memeriksa apakah kode tengah sama dengan target
            return tengah, jumlah_perbandingan  # Mengembalikan indeks dan jumlah perbandingan

        elif data_urut[tengah][0] < target:  # Memeriksa apakah target berada di bagian kanan
            kiri = tengah + 1  # Memindahkan batas kiri ke setelah indeks tengah

        else:  # Dijalankan jika target lebih kecil dari kode tengah
            kanan = tengah - 1  # Memindahkan batas kanan ke sebelum indeks tengah

    return -1, jumlah_perbandingan  # Mengembalikan -1 jika target tidak ditemukan


data_urut = merge_sort_nim(data.copy())  # Mengurutkan salinan data dengan Merge Sort

target = 'TI005'  # Menentukan kode mahasiswa yang dicari
indeks, jumlah = binary_search_nim(data_urut, target)  # Melakukan Binary Search

print("Target:", target)  # Menampilkan kode mahasiswa yang dicari
print("Indeks hasil:", indeks)  # Menampilkan indeks target pada data terurut
print("Jumlah perbandingan:", jumlah)  # Menampilkan jumlah perbandingan pencarian
```

```python
#8. Tampilkan ringkasan hasil yang memuat target, indeks hasil, jumlah perbandingan Linear Search, jumlah perbandingan Binary Search, dan data akhir yang sudah terurut
d = 9  # Menentukan digit terakhir NIM
data = [  # Membuat data mahasiswa
    ['TI001', 'Alya', 78],  # Data mahasiswa pertama
    ['TI002', 'Bima', 80 + d],  # Nilai Bima ditambah digit terakhir NIM
    ['TI003', 'Citra', 67],  # Data mahasiswa ketiga
    ['TI004', 'Danu', 92],  # Data mahasiswa keempat
    ['TI005', 'Eka', 74],  # Data mahasiswa kelima
    ['TI006', 'Fajar', 88],  # Data mahasiswa keenam
    ['TI007', 'Gina', 81],  # Data mahasiswa ketujuh
    ['TI008', 'Hadi', 65 + d],  # Nilai Hadi ditambah digit terakhir NIM
    ['TI009', 'Intan', 91],  # Data mahasiswa kesembilan
    ['TI010', 'Joko', 76],  # Data mahasiswa kesepuluh
    ['TI011', 'Karin', 84],  # Data mahasiswa kesebelas
    ['TI012', 'Lukman', 70 + d]  # Nilai Lukman ditambah digit terakhir NIM
]

def linear_search_nim(data, target):  # Membuat fungsi Linear Search
    jumlah = 0  # Menginisialisasi jumlah perbandingan
    for i in range(len(data)):  # Memeriksa data dari awal
        jumlah += 1  # Menambah jumlah perbandingan
        if data[i][0] == target:  # Memeriksa kode mahasiswa
            return i, jumlah  # Mengembalikan indeks dan jumlah perbandingan
    return -1, jumlah  # Mengembalikan -1 jika target tidak ditemukan

def merge(left, right):  # Membuat fungsi untuk menggabungkan data
    hasil = []  # Menyiapkan list hasil
    i = 0  # Indeks bagian kiri
    j = 0  # Indeks bagian kanan
    while i < len(left) and j < len(right):  # Membandingkan kedua bagian
        if left[i][0] <= right[j][0]:  # Membandingkan kode mahasiswa
            hasil.append(left[i])  # Memasukkan data kiri
            i += 1  # Berpindah ke data kiri berikutnya
        else:  # Jika kode kanan lebih kecil
            hasil.append(right[j])  # Memasukkan data kanan
            j += 1  # Berpindah ke data kanan berikutnya
    while i < len(left):  # Memeriksa sisa bagian kiri
        hasil.append(left[i])  # Memasukkan sisa data kiri
        i += 1  # Berpindah ke data kiri berikutnya
    while j < len(right):  # Memeriksa sisa bagian kanan
        hasil.append(right[j])  # Memasukkan sisa data kanan
        j += 1  # Berpindah ke data kanan berikutnya
    return hasil  # Mengembalikan hasil penggabungan

def merge_sort_nim(data):  # Membuat fungsi Merge Sort rekursif
    if len(data) <= 1:  # Memeriksa kondisi dasar
        return data  # Mengembalikan data yang sudah terurut
    tengah = len(data) // 2  # Menentukan posisi tengah
    left = merge_sort_nim(data[:tengah])  # Mengurutkan bagian kiri
    right = merge_sort_nim(data[tengah:])  # Mengurutkan bagian kanan
    return merge(left, right)  # Menggabungkan kedua bagian

def binary_search_nim(data_urut, target):  # Membuat fungsi Binary Search
    kiri = 0  # Menentukan batas kiri
    kanan = len(data_urut) - 1  # Menentukan batas kanan
    jumlah = 0  # Menginisialisasi jumlah perbandingan
    while kiri <= kanan:  # Mengulang selama rentang pencarian tersedia
        tengah = (kiri + kanan) // 2  # Menghitung indeks tengah
        jumlah += 1  # Menambah jumlah perbandingan
        if data_urut[tengah][0] == target:  # Memeriksa kode pada posisi tengah
            return tengah, jumlah  # Mengembalikan indeks dan jumlah perbandingan
        elif data_urut[tengah][0] < target:  # Jika target lebih besar
            kiri = tengah + 1  # Menggeser batas kiri
        else:  # Jika target lebih kecil
            kanan = tengah - 1  # Menggeser batas kanan
    return -1, jumlah  # Mengembalikan -1 jika target tidak ditemukan

target_index = d % len(data)  # Menghitung indeks target berdasarkan digit terakhir NIM
target_nim = data[target_index][0]  # Menentukan kode mahasiswa target
indeks_linear, jumlah_linear = linear_search_nim(data, target_nim)  # Menjalankan Linear Search
data_urut = merge_sort_nim(data.copy())  # Mengurutkan data menggunakan Merge Sort
indeks_binary, jumlah_binary = binary_search_nim(data_urut, target_nim)  # Menjalankan Binary Search

print("RINGKASAN HASIL PENCARIAN")  # Menampilkan judul ringkasan
print("Target:", target_nim)  # Menampilkan kode target
print("Indeks hasil Linear Search:", indeks_linear)  # Menampilkan indeks Linear Search
print("Jumlah perbandingan Linear Search:", jumlah_linear)  # Menampilkan perbandingan Linear Search
print("Indeks hasil Binary Search:", indeks_binary)  # Menampilkan indeks Binary Search
print("Jumlah perbandingan Binary Search:", jumlah_binary)  # Menampilkan perbandingan Binary Search
print("\nDATA AKHIR SETELAH DIURUTKAN")  # Menampilkan judul data terurut
for mahasiswa in data_urut:  # Memeriksa setiap data yang sudah terurut
    print(mahasiswa)  # Menampilkan data mahasiswa
```

```python
#9. Pada Text cell, jelaskan: kompleksitas Linear Search, Binary Search, dan Merge Sort; kebutuhan ruang tambahan Merge Sort; alasan Binary Search memerlukan data terurut; serta kapan biaya sorting layak dibayar sebelum melakukan pencarian.

 ### Analisis Kompleksitas Algoritma
1. Kompleksitas Linear Search
Linear Search memiliki kompleksitas waktu O(n) pada kasus terburuk karena harus memeriksa data satu per satu sampai target ditemukan atau seluruh data selesai diperiksa. Jika target berada di awal, pencarian dapat selesai lebih cepat.

2. Kompleksitas Binary Search
Binary Search memiliki kompleksitas waktu O(log n) karena setiap langkah membagi area pencarian menjadi dua bagian. Algoritma ini lebih efisien untuk data berjumlah besar, tetapi data harus sudah terurut.

3. Kompleksitas Merge Sort
Merge Sort memiliki kompleksitas waktu O(n log n) pada kasus terbaik, rata-rata, maupun terburuk. Data dibagi menjadi beberapa bagian secara rekursif, kemudian digabungkan kembali dalam keadaan terurut.

4. Kebutuhan Ruang Tambahan Merge Sort
Merge Sort membutuhkan ruang tambahan O(n) untuk menampung data sementara selama proses penggabungan agar hasilnya terurut.

5. Alasan Binary Search Memerlukan Data Terurut
Binary Search memerlukan data terurut karena algoritma menentukan apakah pencarian dilanjutkan ke bagian kiri atau kanan berdasarkan perbandingan dengan data tengah. Jika data belum terurut, target bisa tidak ditemukan meskipun sebenarnya ada.

6. Kapan Biaya Sorting Layak Dibayar?
Biaya sorting layak dibayar ketika data akan dicari berkali-kali, terutama jika jumlah datanya besar. Pengurutan membutuhkan waktu O(n log n), tetapi setelah data terurut, setiap pencarian Binary Search hanya membutuhkan waktu O(log n). Jika pencarian hanya dilakukan satu kali pada data kecil, Linear Search biasanya lebih praktis karena tidak memerlukan pengurutan terlebih dahulu.
```
