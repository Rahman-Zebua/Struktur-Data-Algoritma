```python
#1. Implementasikan seluruh class yang diwajibkan dan uji operasi dasar masing masing class sebelum menjalankan simulasi utama.

# Class Node untuk menyimpan data dan referensi node berikutnya
class Node:
    def __init__(self, data):  # Inisialisasi node
        self.data = data  # Menyimpan data
        self.next = None  # Referensi ke node berikutnya

# Class LinkedQueue berbasis singly linked list
class LinkedQueue:
    def __init__(self):  # Inisialisasi Queue
        self.front = None  # Referensi node paling depan
        self.rear = None  # Referensi node paling belakang
        self.count = 0  # Jumlah data dalam Queue

    def enqueue(self, data):  # Menambahkan data ke belakang
        node_baru = Node(data)  # Membuat node baru
        if self.is_empty():  # Memeriksa Queue kosong
            self.front = node_baru  # Node baru menjadi front
            self.rear = node_baru  # Node baru menjadi rear
        else:
            self.rear.next = node_baru  # Menghubungkan node terakhir ke node baru
            self.rear = node_baru  # Memindahkan rear ke node baru
        self.count += 1  # Menambah jumlah data

    def enqueue_front(self, data):  # Menambahkan data ke depan
        node_baru = Node(data)  # Membuat node baru
        if self.is_empty():  # Memeriksa Queue kosong
            self.front = node_baru  # Node baru menjadi front
            self.rear = node_baru  # Node baru menjadi rear
        else:
            node_baru.next = self.front  # Menghubungkan node baru ke front lama
            self.front = node_baru  # Memindahkan front ke node baru
        self.count += 1  # Menambah jumlah data

    def dequeue(self):  # Mengambil dan menghapus data paling depan
        if self.is_empty():  # Memeriksa Queue kosong
            print("Queue kosong, tidak ada data yang dapat dilayani.")  # Pesan aman
            return None  # Menghentikan operasi tanpa error
        data = self.front.data  # Mengambil data paling depan
        self.front = self.front.next  # Memindahkan front ke node berikutnya
        self.count -= 1  # Mengurangi jumlah data
        if self.front is None:  # Memeriksa apakah Queue menjadi kosong
            self.rear = None  # Mengosongkan rear
        return data  # Mengembalikan data yang diambil

    def peek(self):  # Melihat data paling depan
        if self.is_empty():  # Memeriksa Queue kosong
            print("Queue kosong, tidak ada data di bagian depan.")  # Pesan aman
            return None  # Tidak ada data untuk dikembalikan
        return self.front.data  # Mengembalikan data depan

    def is_empty(self):  # Memeriksa apakah Queue kosong
        return self.count == 0  # True jika jumlah data nol

    def size(self):  # Menghitung jumlah data
        return self.count  # Mengembalikan jumlah data

    def display(self):  # Menampilkan seluruh data Queue
        data_queue = []  # Menyiapkan daftar untuk tampilan
        current = self.front  # Mulai dari node paling depan
        while current is not None:  # Menelusuri semua node
            data_queue.append(current.data)  # Menambahkan data ke daftar tampilan
            current = current.next  # Berpindah ke node berikutnya
        print("Queue:", data_queue)  # Menampilkan isi Queue

# Class Stack menggunakan Python list sebagai penyimpanan internal
class Stack:
    def __init__(self):  # Inisialisasi Stack
        self.items = []  # Menyimpan data Stack

    def push(self, data):  # Menambahkan data ke atas Stack
        self.items.append(data)  # Menambahkan data ke akhir list

    def pop(self):  # Mengambil dan menghapus data paling atas
        if self.is_empty():  # Memeriksa Stack kosong
            print("Stack kosong, tidak ada data yang dapat diambil.")  # Pesan aman
            return None  # Menghentikan operasi tanpa error
        return self.items.pop()  # Mengambil data paling atas

    def peek(self):  # Melihat data paling atas
        if self.is_empty():  # Memeriksa Stack kosong
            print("Stack kosong, tidak ada data paling atas.")  # Pesan aman
            return None  # Tidak ada data untuk dikembalikan
        return self.items[-1]  # Mengembalikan data paling atas

    def is_empty(self):  # Memeriksa apakah Stack kosong
        return len(self.items) == 0  # True jika Stack kosong

    def size(self):  # Menghitung jumlah data Stack
        return len(self.items)  # Mengembalikan jumlah data

    def display(self):  # Menampilkan seluruh data Stack
        print("Stack (bawah ke atas):", self.items)  # Menampilkan isi Stack

# Pengujian operasi dasar LinkedQueue
print("=== PENGUJIAN LINKEDQUEUE ===")
queue = LinkedQueue()  # Membuat objek Queue
print("Queue kosong:", queue.is_empty())  # Menguji is_empty
queue.enqueue("MHS-901")  # Menambahkan data ke belakang
queue.enqueue("MHS-902")  # Menambahkan data ke belakang
queue.enqueue_front("MHS-903")  # Menambahkan data ke depan
queue.display()  # Menguji display
print("Data depan:", queue.peek())  # Menguji peek
print("Jumlah data:", queue.size())  # Menguji size
print("Data yang diambil:", queue.dequeue())  # Menguji dequeue
queue.display()  # Menampilkan Queue setelah dequeue
print("Jumlah data sekarang:", queue.size())  # Memeriksa jumlah terbaru
print("Data yang diambil:", queue.dequeue())  # Menguji dequeue berikutnya
print("Data yang diambil:", queue.dequeue())  # Mengosongkan Queue
print("Queue kosong:", queue.is_empty())  # Memeriksa Queue kosong
queue.dequeue()  # Menguji dequeue saat Queue kosong

# Pengujian operasi dasar Stack
print("\n=== PENGUJIAN STACK ===")
stack = Stack()  # Membuat objek Stack
print("Stack kosong:", stack.is_empty())  # Menguji is_empty
stack.push("MHS-901")  # Menambahkan data ke Stack
stack.push("MHS-902")  # Menambahkan data ke Stack
stack.push("MHS-903")  # Menambahkan data ke Stack
stack.display()  # Menguji display
print("Data paling atas:", stack.peek())  # Menguji peek
print("Jumlah data:", stack.size())  # Menguji size
print("Data yang diambil:", stack.pop())  # Menguji pop
stack.display()  # Menampilkan Stack setelah pop
print("Jumlah data sekarang:", stack.size())  # Memeriksa jumlah terbaru
print("Data yang diambil:", stack.pop())  # Menguji pop berikutnya
print("Data yang diambil:", stack.pop())  # Mengosongkan Stack
print("Stack kosong:", stack.is_empty())  # Memeriksa Stack kosong
stack.pop()  # Menguji pop saat Stack kosong
```

```python
#2. Jalankan 12 operasi pada tabel secara berurutan. Setelah setiap langkah, tampilkan isi Queue dari front ke rear dan isi Stack riwayat.

# Class Node untuk menyimpan data dan referensi node berikutnya
class Node:
    def __init__(self, data):  # Inisialisasi node
        self.data = data  # Menyimpan data
        self.next = None  # Referensi ke node berikutnya

# Class LinkedQueue berbasis singly linked list
class LinkedQueue:
    def __init__(self):  # Inisialisasi Queue
        self.front = None  # Referensi node paling depan
        self.rear = None  # Referensi node paling belakang
        self.count = 0  # Jumlah data dalam Queue

    def enqueue(self, data):  # Menambahkan data ke belakang Queue
        node_baru = Node(data)  # Membuat node baru
        if self.is_empty():  # Memeriksa apakah Queue kosong
            self.front = node_baru  # Node baru menjadi front
            self.rear = node_baru  # Node baru menjadi rear
        else:
            self.rear.next = node_baru  # Menghubungkan node terakhir ke node baru
            self.rear = node_baru  # Memindahkan rear ke node baru
        self.count += 1  # Menambah jumlah data

    def enqueue_front(self, data):  # Menambahkan data ke depan Queue
        node_baru = Node(data)  # Membuat node baru
        if self.is_empty():  # Memeriksa apakah Queue kosong
            self.front = node_baru  # Node baru menjadi front
            self.rear = node_baru  # Node baru menjadi rear
        else:
            node_baru.next = self.front  # Menghubungkan node baru ke front lama
            self.front = node_baru  # Memindahkan front ke node baru
        self.count += 1  # Menambah jumlah data

    def dequeue(self):  # Mengambil dan menghapus data paling depan
        if self.is_empty():  # Memeriksa apakah Queue kosong
            print("Queue kosong, tidak ada data yang dapat dilayani.")  # Pesan aman
            return None  # Mengakhiri operasi dengan aman
        data = self.front.data  # Mengambil data paling depan
        self.front = self.front.next  # Memindahkan front ke node berikutnya
        self.count -= 1  # Mengurangi jumlah data
        if self.front is None:  # Memeriksa apakah Queue menjadi kosong
            self.rear = None  # Mengosongkan rear
        return data  # Mengembalikan data yang diambil

    def peek(self):  # Melihat data paling depan
        if self.is_empty():  # Memeriksa apakah Queue kosong
            print("Queue kosong, tidak ada data di bagian depan.")  # Pesan aman
            return None  # Tidak ada data
        return self.front.data  # Mengembalikan data depan

    def is_empty(self):  # Memeriksa apakah Queue kosong
        return self.count == 0  # True jika jumlah data nol

    def size(self):  # Menghitung jumlah data Queue
        return self.count  # Mengembalikan jumlah data

    def display(self):  # Menampilkan isi Queue dari front ke rear
        data_queue = []  # Menyiapkan daftar tampilan
        current = self.front  # Mulai dari front
        while current is not None:  # Menelusuri seluruh node
            data_queue.append(current.data)  # Menambahkan data ke tampilan
            current = current.next  # Berpindah ke node berikutnya
        print(data_queue)  # Menampilkan isi Queue

# Class Stack menggunakan Python list
class Stack:
    def __init__(self):  # Inisialisasi Stack
        self.items = []  # Penyimpanan internal Stack

    def push(self, data):  # Menambahkan data ke atas Stack
        self.items.append(data)  # Menambahkan data ke akhir list

    def pop(self):  # Mengambil dan menghapus data paling atas
        if self.is_empty():  # Memeriksa apakah Stack kosong
            print("Stack kosong, tidak ada data yang dapat diambil.")  # Pesan aman
            return None  # Mengakhiri operasi dengan aman
        return self.items.pop()  # Mengambil data paling atas

    def peek(self):  # Melihat data paling atas
        if self.is_empty():  # Memeriksa apakah Stack kosong
            print("Stack kosong, tidak ada data paling atas.")  # Pesan aman
            return None  # Tidak ada data
        return self.items[-1]  # Mengembalikan data paling atas

    def is_empty(self):  # Memeriksa apakah Stack kosong
        return len(self.items) == 0  # True jika Stack kosong

    def size(self):  # Menghitung jumlah data Stack
        return len(self.items)  # Mengembalikan jumlah data

    def display(self):  # Menampilkan Stack dari bawah ke atas
        print(self.items)  # Menampilkan isi Stack

# Membentuk ID layanan berdasarkan digit terakhir NIM
d = 9  # Digit terakhir NIM
A = f'MHS-{d}01'  # ID layanan A
B = f'MHS-{d}02'  # ID layanan B
C = f'MHS-{d}03'  # ID layanan C
D = f'MHS-{d}04'  # ID layanan D
E = f'MHS-{d}05'  # ID layanan E
F = f'MHS-{d}06'  # ID layanan F

# Membuat Queue dan Stack untuk simulasi
queue = LinkedQueue()  # Membuat antrean layanan
riwayat = Stack()  # Membuat Stack riwayat layanan

# Menjalankan satu operasi dan menampilkan kondisi struktur data
def jalankan_operasi(nomor, operasi, data=None):  # Fungsi simulasi
    print(f"\nLangkah {nomor}: {operasi}")  # Menampilkan nomor dan operasi

    if operasi == "DATANG":  # Memeriksa operasi kedatangan
        queue.enqueue(data)  # Memasukkan layanan ke belakang Queue
        print(f"Layanan datang: {data}")  # Menampilkan layanan yang datang

    elif operasi == "LAYANI":  # Memeriksa operasi pelayanan
        layanan = queue.dequeue()  # Mengambil layanan paling depan
        if layanan is not None:  # Memastikan ada layanan yang diambil
            riwayat.push(layanan)  # Menyimpan layanan ke Stack riwayat
            print(f"Layanan diproses: {layanan}")  # Menampilkan layanan diproses

    elif operasi == "UNDO":  # Memeriksa operasi pembatalan layanan terakhir
        if riwayat.is_empty():  # Memeriksa apakah riwayat kosong
            print("Riwayat kosong, tidak ada layanan untuk di-UNDO.")  # Pesan aman
        else:
            layanan = riwayat.pop()  # Mengambil riwayat paling atas
            queue.enqueue_front(layanan)  # Mengembalikan layanan ke depan Queue
            print(f"UNDO layanan: {layanan}")  # Menampilkan layanan yang dikembalikan

    print("Queue (front ke rear):", end=" ")  # Label isi Queue
    queue.display()  # Menampilkan seluruh antrean
    print("Stack riwayat (bawah ke atas):", end=" ")  # Label Stack riwayat
    riwayat.display()  # Menampilkan seluruh riwayat

# Menjalankan 12 operasi sesuai urutan tabel soal
jalankan_operasi(1, "DATANG", A)  # Operasi 1
jalankan_operasi(2, "DATANG", B)  # Operasi 2
jalankan_operasi(3, "DATANG", C)  # Operasi 3
jalankan_operasi(4, "LAYANI")  # Operasi 4
jalankan_operasi(5, "DATANG", D)  # Operasi 5
jalankan_operasi(6, "LAYANI")  # Operasi 6
jalankan_operasi(7, "DATANG", E)  # Operasi 7
jalankan_operasi(8, "UNDO")  # Operasi 8
jalankan_operasi(9, "LAYANI")  # Operasi 9
jalankan_operasi(10, "LAYANI")  # Operasi 10
jalankan_operasi(11, "UNDO")  # Operasi 11
jalankan_operasi(12, "DATANG", F)  # Operasi 12
```

```python
#5. Jelaskan dengan contoh dari simulasi mengapa Queue mengikuti FIFO dan Stack mengikuti LIFO.
Queue mengikuti FIFO (First In, First Out), artinya data yang pertama masuk akan menjadi data pertama yang keluar.
Contohnya pada simulasi, MHS-901 masuk pada langkah 1 dan MHS-902 masuk pada langkah 2. Ketika operasi LAYANI dilakukan pada langkah 4, MHS-901 dilayani terlebih dahulu karena berada di depan Queue.
Setelah itu, MHS-902 dilayani pada langkah 6.

Stack mengikuti LIFO (Last In, First Out), artinya data yang terakhir masuk akan menjadi data pertama yang keluar.
Contohnya, MHS-901 masuk ke Stack riwayat pada langkah 4, kemudian MHS-902 masuk pada langkah 6.
Ketika operasi UNDO dilakukan pada langkah 8, MHS-902 diambil terlebih dahulu dari Stack karena merupakan data terakhir yang masuk.
MHS-902 kemudian dikembalikan ke bagian depan Queue.

Jadi, Queue menggunakan prinsip FIFO untuk melayani antrean secara berurutan, sedangkan Stack
menggunakan prinsip LIFO untuk mengambil riwayat layanan dari yang terakhir dilakukan.
```
