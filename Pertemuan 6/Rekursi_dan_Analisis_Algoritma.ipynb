```python
#6.6. Implementasi Factorial Rekursif
def factorial(n):
  if n < 0:
    raise ValueError('n harus >= 0')
  if n == 0:
    return 1
  return n * factorial(n - 1)
print(factorial(4))
```

```python
#6.6.2 Menguji Beberapa Input
for n in [0, 1, 3, 5, 7]:
  print(n, '!=', factorial(n))
```

```python
#6.7. Melihat Urutan Pemanggilan dengan Trace
def factorial_trace(n, depth=0):
  indent = ' ' * depth
  print(indent + f'Masuk factorial({n})')
  if n == 0:
    print(indent + 'Base case -> return 1')
    return 1
  hasil = n * factorial_trace(n - 1, depth + 1)
  print(indent + f'Keluar factorial({n}) -> {hasil}')
  return hasil
factorial_trace(4)
```

```python
#6.3 Perbandingan alur factorial rekursif dan iteratif
def factorial_iteratif(n):
  if n < 0:
    raise ValueError('n harus >= 0')
  hasil = 1
  for i in range(2, n + 1):
    hasil *= i
  return hasil
print(factorial_iteratif(4))
```

```python
#6.16. Fibonacci Rekursif Naif
 #Deret Fibonacci memiliki definisi F(0)=0, F(1)=1, dan F(n)=F(n−1)+F(n−2). Definisi ini
 #terlihat sangat cocok untuk rekursi, tetapi implementasi langsung melakukan banyak
 #perhitungan submasalah yang sama berulang kali.
def fibonacci(n):
  if n < 0:
    raise ValueError('n harus >= 0')
  if n <= 1:
    return n
  return fibonacci(n - 1) + fibonacci(n - 2)
print(fibonacci(6))
```

```python
#6.17. Mengurangi Perhitungan Ulang dengan Memoization
 #Memoization menyimpan hasil submasalah yang sudah pernah dihitung. Ketika submasalah
 #yang sama muncul kembali, fungsi mengambil nilai dari penyimpanan dan tidak
 #menghitungnya dari awal.
def fibonacci_memo(n, memo=None):
  if memo is None:
    memo = {}
  if n < 0:
    raise ValueError('n harus >= 0')
  if n <= 1:
    return n
  if n in memo:
    return memo[n]
  memo[n] = fibonacci_memo(n - 1, memo) + fibonacci_memo(n
- 2, memo)
  return memo[n]
print(fibonacci_memo(30))
```

```python
#6.17.1 Alternatif dengan functools.lru_cache
from functools import lru_cache
@lru_cache(maxsize=None)
def fib_cached(n):
  if n <= 1:
    return n
  return fib_cached(n - 1) + fib_cached(n - 2)
print(fib_cached(30))
```

```python
#Gambar 6.7 Penyempitan ruang pencarian pada recursive binary search
def binary_search_recursive(data, target, low=0, high=None):
    if high is None:
        high = len(data) - 1

    if low > high:
        return -1

    mid = (low + high) // 2

    if data[mid] == target:
        return mid
    elif target < data[mid]:
        return binary_search_recursive(data, target, low, mid - 1)
    else:
        return binary_search_recursive(data, target, mid + 1, high)


data = [3, 7, 12, 18, 24, 31, 45, 59, 63, 80, 91, 105]
print(binary_search_recursive(data, 63))
```

```python
#6.19. Menghitung Operasi untuk Memahami Pertumbuhan
 #Sebelum menggunakan waktu eksekusi, kita dapat menghitung banyaknya operasi penting.
 #Pendekatan ini lebih stabil karena tidak bergantung pada kecepatan komputer.
def linear_search_count(data, target):
    perbandingan = 0
    for i, nilai in enumerate(data):
        perbandingan += 1
        if nilai == target:
            return i, perbandingan
    return -1, perbandingan


for n in [10, 100, 1000]:
    data = list(range(n))
    _, count = linear_search_count(data, -1)
    print(n, "elemen ->", count, "perbandingan")
```

```python
#6.20. Eksperimen Runtime dengan timeit
import timeit


def factorial_iteratif(n):
    hasil = 1
    for i in range(2, n + 1):
        hasil *= i
    return hasil


def factorial_rekursif(n):
    if n == 0:
        return 1
    return n * factorial_rekursif(n - 1)


for n in [10, 100, 500]:
    t_iter = timeit.timeit(lambda: factorial_iteratif(n), number=1000)
    t_rec = timeit.timeit(lambda: factorial_rekursif(n), number=1000)
    print(n, t_iter, t_rec)
```

```python
#6.24. Mini Studi Kasus — Menjumlahkan Data dengan Divide-and-Conquer Rekursif
 #Kita akan membuat fungsi yang menjumlahkan data dengan membagi list menjadi dua
 #bagian. Contoh ini bukan cara paling efisien untuk fungsi sum di Python, tetapi baik untuk
 #memahami rekursi yang membagi masalah menjadi submasalah.
def recursive_sum(data):
    if len(data) == 0:
        return 0
    if len(data) == 1:
        return data[0]

    mid = len(data) // 2
    kiri = recursive_sum(data[:mid])
    kanan = recursive_sum(data[mid:])

    return kiri + kanan


nilai = [12, 8, 15, 10, 5, 20, 7, 13]
print("Total:", recursive_sum(nilai))
```
