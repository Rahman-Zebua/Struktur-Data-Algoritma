```python
#Bandingkan `preorder_list()` dan `preorder_iterative()`. Pastikan kedua fungsi menghasilkan
#list yang sama menggunakan `assert`. Jelaskan perbedaan penggunaan call stack implisit dan
#stack eksplisit.
class Node:  # Mendefinisikan class Node untuk satu simpul pada tree
    def __init__(self, data):  # Konstruktor saat membuat objek Node baru
        self.data = data  # Menyimpan nilai/data pada node
        self.left = None  # Menginisialisasi cabang anak kiri sebagai None
        self.right = None  # Menginisialisasi cabang anak kanan sebagai None


root = Node("A")  # Membuat node root bernilai "A"
root.left = Node("B")  # Menghubungkan "B" sebagai anak kiri dari "A"
root.right = Node("C")  # Menghubungkan "C" sebagai anak kanan dari "A"
root.left.left = Node("D")  # Menghubungkan "D" sebagai anak kiri dari "B"
root.left.right = Node("F")  # Menghubungkan "F" sebagai anak kanan dari "B"
root.right.right = Node("G")  # Menghubungkan "G" sebagai anak kanan dari "C"


def preorder_list(node):  # Fungsi preorder rekursif yang mengembalikan list (Root -> Left -> Right)
    if node is None:  # Base case: jika node kosong (None)
        return []  # Mengembalikan list kosong
    return [node.data] + preorder_list(node.left) + preorder_list(node.right)  # Ambil data node dulu, lalu kiri, lalu kanan


def preorder_iterative(root_node):  # Fungsi preorder iteratif menggunakan stack eksplisit (LIFO)
    if root_node is None:  # Jika tree kosong
        return []  # Mengembalikan list kosong
    hasil = []  # List untuk menyimpan urutan hasil preorder
    stack = [root_node]  # Membuat stack eksplisit dan memasukkan root sebagai elemen awal
    while stack:  # Selama stack belum kosong, lanjutkan proses
        sekarang = stack.pop()  # Mengambil dan menghapus elemen teratas stack (LIFO: Last-In, First-Out)
        hasil.append(sekarang.data)  # Mencatat data node yang sedang dikunjungi ke list hasil
        if sekarang.right is not None:  # Cek apakah ada anak kanan
            stack.append(sekarang.right)  # Dorong anak kanan ke stack lebih dulu agar diproses setelah anak kiri
        if sekarang.left is not None:  # Cek apakah ada anak kiri
            stack.append(sekarang.left)  # Dorong anak kiri ke stack terakhir agar di-pop lebih dulu (LIFO)
    return hasil  # Mengembalikan list hasil traversal preorder


hasil_rekursif = preorder_list(root)  # Menjalankan preorder_list() dan menyimpan hasilnya
hasil_iteratif = preorder_iterative(root)  # Menjalankan preorder_iterative() dan menyimpan hasilnya

assert hasil_rekursif == hasil_iteratif, "Hasil preorder_list dan preorder_iterative tidak sama!"  # Memastikan kedua list identik; jika beda muncul AssertionError

print("Preorder rekursif :", hasil_rekursif)  # Mencetak hasil dari preorder_list()
print("Preorder iteratif :", hasil_iteratif)  # Mencetak hasil dari preorder_iterative()
print("Status assert     : Berhasil (kedua list sama)")  # Konfirmasi jika assert lolos tanpa error
```
