```python
#Membuat Class Node
class Node: #Mendefinisikan tipe objek node.
  def __init__(self, data):#Constructor yang dijalankan ketika Node dibuat.
    self.data = data #Menyimpan nilai node.
    self.left = None #Belum ada child kiri.
    self.right = None #Belum ada child kanan.
```

```python
#9.6. Membangun Binary Tree Secara Manual
root = Node('A')
root.left = Node('B')
root.right = Node('C')
root.left.left = Node('D')
root.left.right = Node('E')
root.right.left = Node('F')
root.right.right = Node('G')
```

```python
print(root.data)
print(root.left.data)
print(root.right.data)
print(root.left.left.data)
```

```python
#Implementasi Rekursif
def preorder(node):
  if node is None:
    return
  print(node.data, end=' ')
  preorder(node.left)
  preorder(node.right)
preorder(root)
```

```python
#9.10. Mengembalikan Hasil Traversal sebagai List
def preorder_list(node):
    if node is None:
        return []
    return [node.data] + preorder_list(node.left) + preorder_list(node.right)

def inorder_list(node):
    if node is None:
        return []
    return inorder_list(node.left) + [node.data] + inorder_list(node.right)

def postorder_list(node):
    if node is None:
        return []
    return postorder_list(node.left) + postorder_list(node.right) + [node.data]

print(preorder_list(root))
print(inorder_list(root))
print(postorder_list(root))
```

```python
#9.11. Preorder Iteratif Menggunakan Stack
def preorder_iterative(root):
    if root is None:
        return []
    result = []
    stack = [root]
    while stack:
        node = stack.pop()
        result.append(node.data)
        if node.right is not None:
            stack.append(node.right)
        if node.left is not None:
            stack.append(node.left)
    return result

print(preorder_iterative(root))
```

```python
#9.12. Level-Order Traversal Menggunakan Queue
from collections import deque

def level_order(root):
    if root is None:
        return []
    result = []
    queue = deque([root])
    while queue:
        node = queue.popleft()
        result.append(node.data)
        if node.left is not None:
            queue.append(node.left)
        if node.right is not None:
            queue.append(node.right)
    return result

print(level_order(root))
```

```python
#9.13. Menghitung Jumlah Node dan Leaf
#Menghitung Seluruh Node
def count_nodes(node):
  if node is None:
    return 0
  return 1 + count_nodes(node.left) + count_nodes(node.right)
print(count_nodes(root))
```

```python
#Menghitung Leaf
def count_leaves(node):
  if node is None:
    return 0
  if node.left is None and node.right is None:
    return 1
  return count_leaves(node.left) + count_leaves(node.right)
print(count_leaves(root))
```

```python
#9.14. Depth dan Height
#Menghitung Height dengan Rekursi
def height(node):
  if node is None:
    return -1
  left_height = height(node.left)
  right_height = height(node.right)
  return 1 + max(left_height, right_height)
print(height(root))
```

```python
#Menghitung Depth Target
def find_depth(node, target, depth=0):
    if node is None:
        return -1
    if node.data == target:
        return depth
    left_result = find_depth(node.left, target, depth + 1)
    if left_result != -1:
        return left_result
    return find_depth(node.right, target, depth + 1)

print(find_depth(root, 'E'))
```

```python
#9.15. Pencarian pada Binary Tree Umum
def search(node, target):
    if node is None:
        return False
    if node.data == target:
        return True
    return search(node.left, target) or search(node.right, target)

print(search(root, 'F'))
print(search(root, 'Z'))
```

```python
#Versi Level-Order
def search_level_order(root, target):
    if root is None:
        return False
    queue = deque([root])
    while queue:
        node = queue.popleft()
        if node.data == target:
            return True
        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)
    return False
```

```python
#Membangun Expression Tree
expr = Node('*')
expr.left = Node('+')
expr.right = Node('-')
expr.left.left = Node(3)
expr.left.right = Node(5)
expr.right.left = Node(10)
expr.right.right = Node(4)
```

```python
#9.18. Mengevaluasi Expression Tree
def evaluate(node):
    if node.left is None and node.right is None:
        return node.data
    left_value = evaluate(node.left)
    right_value = evaluate(node.right)
    if node.data == '+':
        return left_value + right_value
    if node.data == '-':
        return left_value - right_value
    if node.data == '*':
        return left_value * right_value
    if node.data == '/':
        return left_value / right_value
    raise ValueError('Operator tidak dikenali')

print(evaluate(expr))
```

```python
#Membuat Infix dengan Tanda Kurung
def to_infix(node):
  if node.left is None and node.right is None:
    return str(node.data)
  return f'({to_infix(node.left)} {node.data} {to_infix(node.right)})'
print(to_infix(expr))
```

```python
#9.19. Kesalahan Umum dan Debugging
#Teknik Debugging dengan Print
def preorder_debug(node, depth=0):
    if node is None:
        print(' ' * depth + 'None')
        return
    print(' ' * depth + f'Visit {node.data}')
    preorder_debug(node.left, depth + 1)
    preorder_debug(node.right, depth + 1)

preorder_debug(root)
```
