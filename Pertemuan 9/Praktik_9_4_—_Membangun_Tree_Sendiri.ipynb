```python
#Gambar tree
        50          <- Root (Level 0)
       /  \
     30    70       <- Level 1
    /  \     \
  20    40    80    <- Level 2 (Leaf)
```

```python
#Bangun tree dengan class `Node`.
class Node:  # Mendefinisikan class bernama Node untuk merepresentasikan satu simpul/titik pada tree
    def __init__(self, data):  # Konstruktor yang otomatis berjalan saat objek Node baru dibuat dengan nilai data
        self.data = data  # Menyimpan nilai/isi utama pada node ini
        self.left = None  # Menginisialisasi pointer/cabang anak kiri sebagai None (belum punya anak kiri)
        self.right = None  # Menginisialisasi pointer/cabang anak kanan sebagai None (belum punya anak kanan)


root = Node(50)  # Membuat node akar (root) utama tree dengan nilai 50

root.left = Node(30)  # Membuat node bernilai 30 dan menghubungkannya sebagai anak kiri dari root (50)
root.right = Node(70)  # Membuat node bernilai 70 dan menghubungkannya sebagai anak kanan dari root (50)

root.left.left = Node(20)  # Membuat node bernilai 20 sebagai anak kiri dari node 30
root.left.right = Node(40)  # Membuat node bernilai 40 sebagai anak kanan dari node 30

root.right.right = Node(80)  # Membuat node bernilai 80 sebagai anak kanan dari node 70 (anak kiri 70 dibiarkan None)
```

```python
#Tampilkan preorder, inorder, postorder, dan level-order.
from collections import deque  # Mengimpor deque untuk antrean (queue) efisien pada level-order traversal
class Node:  # Mendefinisikan class Node untuk merepresentasikan satu simpul pada tree
    def __init__(self, data):  # Konstruktor yang dipanggil saat membuat objek Node baru
        self.data = data  # Menyimpan nilai/data pada node
        self.left = None  # Menginisialisasi cabang/pointer anak kiri dengan None
        self.right = None  # Menginisialisasi cabang/pointer anak kanan dengan None


root = Node(50)  # Membuat node root dengan nilai 50
root.left = Node(30)  # Membuat node 30 sebagai anak kiri dari root (50)
root.right = Node(70)  # Membuat node 70 sebagai anak kanan dari root (50)
root.left.left = Node(20)  # Membuat node 20 sebagai anak kiri dari node 30
root.left.right = Node(40)  # Membuat node 40 sebagai anak kanan dari node 30
root.right.right = Node(80)  # Membuat node 80 sebagai anak kanan dari node 70


def preorder(node):  # Fungsi traversal preorder dengan urutan: Root -> Left -> Right
    if node is None:  # Base case rekursi: jika node kosong (None)
        return []  # Mengembalikan list kosong
    return [node.data] + preorder(node.left) + preorder(node.right)  # Ambil data node dulu, lalu kiri, lalu kanan


def inorder(node):  # Fungsi traversal inorder dengan urutan: Left -> Root -> Right
    if node is None:  # Base case rekursi: jika node kosong (None)
        return []  # Mengembalikan list kosong
    return inorder(node.left) + [node.data] + inorder(node.right)  # Telusuri kiri dulu, lalu ambil data node, lalu kanan


def postorder(node):  # Fungsi traversal postorder dengan urutan: Left -> Right -> Root
    if node is None:  # Base case rekursi: jika node kosong (None)
        return []  # Mengembalikan list kosong
    return postorder(node.left) + postorder(node.right) + [node.data]  # Telusuri kiri dulu, lalu kanan, terakhir ambil data node


def level_order(root_node):  # Fungsi traversal level-order (Breadth-First Search) per level dari atas ke bawah
    if root_node is None:  # Jika tree kosong (tidak ada root)
        return []  # Mengembalikan list kosong
    hasil = []  # Membuat list untuk menampung urutan nilai node hasil traversal
    antrean = deque([root_node])  # Memasukkan root ke dalam antrean (FIFO: First-In, First-Out)
    while antrean:  # Selama antrean masih memiliki node untuk diproses
        sekarang = antrean.popleft()  # Mengambil dan menghapus node paling depan dari antrean
        hasil.append(sekarang.data)  # Menyimpan nilai node yang sedang dikunjungi ke list hasil
        if sekarang.left is not None:  # Mengecek apakah node saat ini punya anak kiri
            antrean.append(sekarang.left)  # Memasukkan anak kiri ke bagian belakang antrean
        if sekarang.right is not None:  # Mengecek apakah node saat ini punya anak kanan
            antrean.append(sekarang.right)  # Memasukkan anak kanan ke bagian belakang antrean
    return hasil  # Mengembalikan seluruh urutan nilai node secara level-order


print("Preorder   :", preorder(root))  # Menampilkan hasil preorder: [50, 30, 20, 40, 70, 80]
print("Inorder    :", inorder(root))  # Menampilkan hasil inorder: [20, 30, 40, 50, 70, 80]
print("Postorder  :", postorder(root))  # Menampilkan hasil postorder: [20, 40, 30, 80, 70, 50]
print("Level-order:", level_order(root))  # Menampilkan hasil level-order: [50, 30, 70, 20, 40, 80]
```

```python
#Hitung jumlah node dan
def hitung_jumlah_node(node):  # Mendefinisikan fungsi rekursif untuk menghitung total node pada subtree yang diawali node
    if node is None:  # Base case: jika posisi saat ini kosong/tidak ada node
        return 0  # Mengembalikan 0 karena tidak ada node yang dihitung
    return 1 + hitung_jumlah_node(node.left) + hitung_jumlah_node(node.right)  # Menghitung 1 node saat ini ditambah jumlah node anak kiri dan anak kanan


print("Jumlah node:", hitung_jumlah_node(root))  # Menampilkan jumlah total node pada tree (hasilnya: 6)
```

```python
#Hitung height tree.
def hitung_height(node):  # Mendefinisikan fungsi rekursif untuk menghitung tinggi (height) tree berdasarkan jumlah edge
    if node is None:  # Base case: jika node kosong (None), tinggi dihitung -1
        return -1  # Mengembalikan -1 agar node daun (tanpa anak) bernilai tinggi 0
    tinggi_kiri = hitung_height(node.left)  # Menghitung secara rekursif tinggi subtree pada anak kiri
    tinggi_kanan = hitung_height(node.right)  # Menghitung secara rekursif tinggi subtree pada anak kanan
    return 1 + max(tinggi_kiri, tinggi_kanan)  # Mengambil tinggi terbesar di antara kiri/kanan lalu ditambah 1 edge ke node saat ini


print("Height tree:", hitung_height(root))  # Menampilkan tinggi tree ke layar (hasilnya: 2)
```

```python
#Cari nilai F dan nilai X menggunakan fungsi search.
class Node:  # Mendefinisikan class Node untuk mewakili satu simpul pada tree
    def __init__(self, data):  # Konstruktor untuk menginisialisasi objek Node baru
        self.data = data  # Menyimpan nilai/data pada node
        self.left = None  # Menginisialisasi pointer anak kiri sebagai None
        self.right = None  # Menginisialisasi pointer anak kanan sebagai None


root = Node("A")  # Membuat node root dengan nilai "A"
root.left = Node("B")  # Menghubungkan node "B" sebagai anak kiri dari "A"
root.right = Node("C")  # Menghubungkan node "C" sebagai anak kanan dari "A"
root.left.left = Node("D")  # Menghubungkan node "D" sebagai anak kiri dari "B"
root.left.right = Node("F")  # Menghubungkan node "F" sebagai anak kanan dari "B"
root.right.right = Node("G")  # Menghubungkan node "G" sebagai anak kanan dari "C" ("X" sengaja tidak dimasukkan agar terlihat hasil False)


def search(node, target):  # Mendefinisikan fungsi rekursif untuk mencari nilai target di dalam tree
    if node is None:  # Base case: jika node kosong (None), berarti nilai tidak ditemukan di jalur ini
        return False  # Mengembalikan False karena node tidak ada
    if node.data == target:  # Mengecek apakah data pada node saat ini sama dengan target yang dicari
        return True  # Mengembalikan True jika nilai target ditemukan
    return search(node.left, target) or search(node.right, target)  # Mencari ke anak kiri ATAU anak kanan jika belum ketemu


print("Cari 'F':", search(root, "F"))  # Memanggil fungsi search untuk "F" dan mencetak hasilnya (True jika ada)
print("Cari 'X':", search(root, "X"))  # Memanggil fungsi search untuk "X" dan mencetak hasilnya (False jika tidak ada)
```
