```python
#Gambar tree
         K          <- Root (Level 0)
       /   \
      D     M       <- Level 1
    /  \     \
   B    F     Z     <- Level 2 (Leaf)
```

```python
#Bangun tree dengan class `Node`.
class Node:  # Mendefinisikan class bernama Node untuk merepresentasikan satu simpul/titik pada tree
    def __init__(self, data):  # Konstruktor yang otomatis berjalan saat objek Node baru dibuat dengan nilai data
        self.data = data  # Menyimpan nilai/isi utama pada node ini
        self.left = None  # Menginisialisasi pointer/cabang anak kiri sebagai None (belum punya anak kiri)
        self.right = None  # Menginisialisasi pointer/cabang anak kanan sebagai None (belum punya anak kanan)


root = Node('K')  # Membuat node akar (root) utama tree dengan nilai 'K'

root.left = Node('D')  # Membuat node bernilai 'D' dan menghubungkannya sebagai anak kiri dari root ('K')
root.right = Node('M')  # Membuat node bernilai 'M' dan menghubungkannya sebagai anak kanan dari root ('K')

root.left.left = Node('B')  # Membuat node bernilai 'B' sebagai anak kiri dari node 'D'
root.left.right = Node('F')  # Membuat node bernilai 'F' sebagai anak kanan dari node 'D'

root.right.right = Node('Z')  # Membuat node bernilai 'Z' sebagai anak kanan dari node 'M' (anak kiri 'M' dibiarkan None)
print("Root:", root.data)  # Menampilkan nilai root ('K')
print("Child kiri K:", root.left.data)  # Menampilkan nilai anak kiri 'K' ('D')
print("Child kanan K:", root.right.data)  # Menampilkan nilai anak kanan 'K' ('M')
print("Child kiri D:", root.left.left.data)  # Menampilkan nilai anak kiri 'D' ('B')
print("Child kanan D:", root.left.right.data)  # Menampilkan nilai anak kanan 'D' ('F')
print("Child kiri M:", root.right.left)  # Menampilkan anak kiri 'M' (None)
print("Child kanan M:", root.right.right.data)  # Menampilkan nilai anak kanan 'M' ('Z')
```

```python
#Tampilkan preorder, inorder, postorder, dan level-order.
from collections import deque
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


root = Node('K')
root.left = Node('D')
root.right = Node('M')
root.left.left = Node('B')
root.left.right = Node('F')
root.right.right = Node('Z')


def preorder(node):
    if node is None:
        return []
    return [node.data] + preorder(node.left) + preorder(node.right)


def inorder(node):
    if node is None:
        return []
    return inorder(node.left) + [node.data] + inorder(node.right)


def postorder(node):
    if node is None:
        return []
    return postorder(node.left) + postorder(node.right) + [node.data]


def level_order(node):
    if node is None:
        return []
    hasil = []
    antrian = deque([node])
    while antrian:
        sekarang = antrian.popleft()
        hasil.append(sekarang.data)
        if sekarang.left:
            antrian.append(sekarang.left)
        if sekarang.right:
            antrian.append(sekarang.right)
    return hasil


print("Preorder   :", preorder(root))
print("Inorder    :", inorder(root))
print("Postorder  :", postorder(root))
print("Level-order:", level_order(root))
```

```python
#Hitung jumlah node dan leaf.
class Node:  # Mendefinisikan class bernama Node untuk merepresentasikan satu simpul/titik pada tree
    def __init__(self, data):  # Konstruktor yang otomatis berjalan saat objek Node baru dibuat dengan nilai data
        self.data = data  # Menyimpan nilai/isi utama pada node ini
        self.left = None  # Menginisialisasi pointer/cabang anak kiri sebagai None
        self.right = None  # Menginisialisasi pointer/cabang anak kanan sebagai None


root = Node('K')  # Membuat node akar (root) utama tree dengan nilai 'K'

root.left = Node('D')  # Membuat node bernilai 'D' sebagai anak kiri dari root ('K')
root.right = Node('M')  # Membuat node bernilai 'M' sebagai anak kanan dari root ('K')

root.left.left = Node('B')  # Membuat node bernilai 'B' sebagai anak kiri dari node 'D'
root.left.right = Node('F')  # Membuat node bernilai 'F' sebagai anak kanan dari node 'D'

root.right.right = Node('Z')  # Membuat node bernilai 'Z' sebagai anak kanan dari node 'M'


def hitung_node(node):  # Fungsi rekursif untuk menghitung total node
    if node is None:  # Jika kosong, jumlahnya 0
        return 0
    return 1 + hitung_node(node.left) + hitung_node(node.right)  # 1 (node saat ini) + kiri + kanan


def hitung_leaf(node):  # Fungsi rekursif untuk menghitung jumlah leaf (tidak punya anak kiri & kanan)
    if node is None:  # Jika kosong, bukan leaf
        return 0
    if node.left is None and node.right is None:  # Jika tidak punya anak kiri maupun kanan, ini leaf
        return 1
    return hitung_leaf(node.left) + hitung_leaf(node.right)  # Jumlahkan leaf di subtree kiri dan kanan


print("Jumlah node:", hitung_node(root))
print("Jumlah leaf:", hitung_leaf(root))
```

```python
#Hitung height tree.
class Node:  # Mendefinisikan class bernama Node untuk merepresentasikan satu simpul/titik pada tree
    def __init__(self, data):  # Konstruktor yang otomatis berjalan saat objek Node baru dibuat dengan nilai data
        self.data = data  # Menyimpan nilai/isi utama pada node ini
        self.left = None  # Menginisialisasi pointer/cabang anak kiri sebagai None
        self.right = None  # Menginisialisasi pointer/cabang anak kanan sebagai None


root = Node('K')  # Membuat node akar (root) utama tree dengan nilai 'K'

root.left = Node('D')  # Membuat node bernilai 'D' sebagai anak kiri dari root ('K')
root.right = Node('M')  # Membuat node bernilai 'M' sebagai anak kanan dari root ('K')

root.left.left = Node('B')  # Membuat node bernilai 'B' sebagai anak kiri dari node 'D'
root.left.right = Node('F')  # Membuat node bernilai 'F' sebagai anak kanan dari node 'D'

root.right.right = Node('Z')  # Membuat node bernilai 'Z' sebagai anak kanan dari node 'M'


# Menghitung height berdasarkan jumlah edge (root ke leaf = 2, tree kosong/single node = -1 atau 0 sesuai konvensi)
def hitung_height(node):
    if node is None:  # Base case: tree kosong mengembalikan -1 agar leaf punya height 0
        return -1
    left_height = hitung_height(node.left)    # Tinggi subtree kiri
    right_height = hitung_height(node.right)  # Tinggi subtree kanan
    return 1 + max(left_height, right_height) # Ambil maksimum lalu tambah 1 edge ke node saat ini


print("Height tree (berdasarkan edge):", hitung_height(root))
```

```python
#Cari nilai F dan nilai X menggunakan fungsi search.
class Node:  # Mendefinisikan class bernama Node untuk merepresentasikan satu simpul/titik pada tree
    def __init__(self, data):  # Konstruktor yang otomatis berjalan saat objek Node baru dibuat dengan nilai data
        self.data = data  # Menyimpan nilai/isi utama pada node ini
        self.left = None  # Menginisialisasi pointer/cabang anak kiri sebagai None
        self.right = None  # Menginisialisasi pointer/cabang anak kanan sebagai None


root = Node('K')  # Membuat node akar (root) utama tree dengan nilai 'K'

root.left = Node('D')  # Membuat node bernilai 'D' sebagai anak kiri dari root ('K')
root.right = Node('M')  # Membuat node bernilai 'M' sebagai anak kanan dari root ('K')

root.left.left = Node('B')  # Membuat node bernilai 'B' sebagai anak kiri dari node 'D'
root.left.right = Node('F')  # Membuat node bernilai 'F' sebagai anak kanan dari node 'D'

root.right.right = Node('Z')  # Membuat node bernilai 'Z' sebagai anak kanan dari node 'M'


def search(node, target):  # Fungsi rekursif untuk mencari nilai target di seluruh binary tree
    if node is None:  # Base case: jika sampai di ujung cabang kosong, target tidak ditemukan
        return False
    if node.data == target:  # Jika data node saat ini sama dengan target, kembalikan True
        return True
    # Cari di subtree kiri atau subtree kanan
    return search(node.left, target) or search(node.right, target)


print("Cari 'F':", search(root, 'F'))
print("Cari 'X':", search(root, 'X'))
```
