```python
#2. Jalankan push untuk data A, B, C.
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
class LinkedStack:
    def __init__(self):
        self.top = None
    def is_empty(self):
        return self.top is None
    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
stack = LinkedStack()
stack.push("A")
stack.push("B")
stack.push("C")
current = stack.top
while current is not None:
    print(current.data)
    current = current.next
```

```python
#4. Jalankan pop sekali dan gambarkan kembali strukturnya.
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
class LinkedStack:
    def __init__(self):
        self.top = None
    def is_empty(self):
        return self.top is None
    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
    def pop(self):
        if self.is_empty():
            return None
        data = self.top.data
        self.top = self.top.next
        return data
    def display(self):
        current = self.top
        while current is not None:
            print(current.data)
            current = current.next
stack = LinkedStack()
stack.push("A")
stack.push("B")
stack.push("C")
print("Data yang di-pop:", stack.pop())
print("Isi Stack setelah pop:")
stack.display()
```

```python
#5. Uji peek dan size.
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
class LinkedStack:
    def __init__(self):
        self.top = None
    def is_empty(self):
        return self.top is None
    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
    def pop(self):
        if self.is_empty():
            return None
        data = self.top.data
        self.top = self.top.next
        return data
    def peek(self):
        if self.is_empty():
            return None
        return self.top.data
    def size(self):
        count = 0
        current = self.top
        while current is not None:
            count += 1
            current = current.next
        return count
stack = LinkedStack()
stack.push("A")
stack.push("B")
stack.push("C")
stack.pop()
print("Data paling atas:", stack.peek())
print("Jumlah data dalam Stack:", stack.size())
```

```python
#6. Bandingkan kode push/pop pada LinkedStack dengan insert/delete pada awal Singly Linked List Pertemuan 3.

Kode `push()` dan `pop()` pada LinkedStack memiliki cara kerja yang hampir sama dengan `insert()` dan `delete()` pada awal Singly Linked List. Pada LinkedStack, `push()` menambahkan node baru di bagian atas Stack, sedangkan `pop()` menghapus node paling atas. Pada Singly Linked List, `insert()` menambahkan node di awal list, sedangkan `delete()` menghapus node pertama.

Perbedaannya terletak pada tujuan penggunaannya. LinkedStack menerapkan prinsip LIFO (Last In, First Out), yaitu data yang terakhir masuk akan keluar terlebih dahulu. Sementara itu, Singly Linked List digunakan untuk menyimpan dan mengelola data secara berurutan, sehingga operasi penambahan dan penghapusan dapat disesuaikan dengan kebutuhan program.
```
