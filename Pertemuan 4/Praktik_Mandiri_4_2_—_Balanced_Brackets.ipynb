```python
#1. Implementasikan fungsi is_balanced().
def is_balanced(expression):
    stack = []
    pasangan = {')': '(', ']': '[', '}': '{'}
    for char in expression:
        if char in '([{':
            stack.append(char)
        elif char in ')]}':
            if not stack or stack.pop() != pasangan[char]:
                return False
    return len(stack) == 0
```

```python
#2. Uji ekspresi: (a+b), ([{}]), ([)], ((a+b), dan {a+[b*c]}.
ekspresi = ["(a+b)", "([{}])", "([)]", "((a+b)", "{a+[b*c]}"]
for x in ekspresi:
    print(x, "->", is_balanced(x))
```

```python
#4. Modifikasi fungsi agar mengabaikan seluruh karakter selain enam simbol bracket.
def is_balanced(expression):
    stack = []
    pasangan = {')': '(', ']': '[', '}': '{'}
    for char in expression:
        if char in '([{':
            stack.append(char)
        elif char in ')]}':
            if not stack or stack.pop() != pasangan[char]:
                return False
    return len(stack) == 0
print(is_balanced("(a+b)"))
print(is_balanced("([{}])"))
print(is_balanced("([)]"))
print(is_balanced("((a+b)"))
print(is_balanced("{a+[b*c]}"))
```

```python
#5. Jelaskan mengapa hanya membandingkan jumlah pembuka dan penutup tidak cukup.
Membandingkan jumlah tanda kurung buka dan tutup saja tidak cukup karena kita juga harus memeriksa urutan dan kecocokan pasangan kurung. Contohnya, ekspresi `([)]` memiliki jumlah kurung buka dan tutup yang sama, tetapi urutannya salah sehingga tidak seimbang. Oleh karena itu, Stack digunakan untuk memastikan setiap tanda kurung tutup cocok dengan tanda kurung buka terakhir yang belum ditutup.
```
