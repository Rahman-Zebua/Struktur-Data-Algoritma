```python
#1.Buat class Stack menggunakan Python list seperti pada modul, tetapi ketik ulang tanpa copy-paste.
class Stack:
    def __init__(self):
        # Inisialisasi list kosong untuk menampung elemen stack
        self.items = []

    def is_empty(self):
        # Memeriksa apakah stack kosong
        return len(self.items) == 0

    def push(self, item):
        # Menambahkan elemen baru ke bagian atas stack
        self.items.append(item)

    def pop(self):
        # Menghapus dan mengembalikan elemen teratas dari stack
        if not self.is_empty():
            return self.items.pop()
        raise IndexError("pop from empty stack")

    def peek(self):
        # Melihat elemen teratas tanpa menghapusnya
        if not self.is_empty():
            return self.items[-1]
        raise IndexError("peek from empty stack")

    def size(self):
        # Mengembalikan jumlah elemen dalam stack
        return len(self.items)


# --- Contoh Penggunaan ---
s = Stack()

# Menambahkan data (push)
s.push("A")
s.push("B")
s.push("C")

print("Elemen teratas (peek):", s.peek())  # Output: C
print("Ukuran stack:", s.size())            # Output: 3

# Mengambil data (pop)
print("Pop elemen:", s.pop())               # Output: C
print("Elemen teratas sekarang:", s.peek())  # Output: B
print("Apakah stack kosong?:", s.is_empty()) # Output: False
```

```python
#2.Tambahkan data 10, 20, 30, 40 menggunakan push.

# Inisialisasi objek Stack baru
s = Stack()

# Menambahkan data 10, 20, 30, 40
s.push(10)
s.push(20)
s.push(30)
s.push(40)

# Menampilkan isi stack dan elemen teratas
print("Isi stack saat ini:", s.items)
print("Elemen teratas (peek):", s.peek())
```

```python
#3.Tampilkan top dan size.

# Menampilkan elemen teratas (top) dan ukuran stack (size)
print("Elemen teratas (top):", s.peek())
print("Ukuran stack (size):", s.size())
```

```python
#4.Lakukan pop dua kali dan catat urutan nilai yang keluar.

# Melakukan pop pertama dan kedua
pop1 = s.pop()
pop2 = s.pop()

# Menampilkan hasil pop
print("Hasil pop pertama :", pop1)
print("Hasil pop kedua   :", pop2)
print("Urutan nilai keluar:", [pop1, pop2])

# Menampilkan sisa isi stack
print("Sisa isi stack saat ini:", s.items)
```

```python
#5.Tambahkan data 50 dan 60.

# Menambahkan data 50 dan 60 menggunakan push
s.push(50)
s.push(60)

# Menampilkan isi stack terbaru dan elemen teratas
print("Isi stack saat ini:", s.items)
print("Elemen teratas (peek):", s.peek())
```

```python
#6.Buat loop yang mengeluarkan semua elemen sampai Stack kosong.

# Loop untuk mengeluarkan semua elemen sampai Stack kosong
print("Mengeluarkan elemen satu per satu:")
while not s.is_empty():
    elemen = s.pop()
    print("Elemen dikeluarkan:", elemen)

# Verifikasi bahwa stack sudah kosong
print("Apakah stack kosong?:", s.is_empty())
print("Ukuran stack sekarang:", s.size())
```

```python
#7.Uji pop ketika Stack kosong dan jelaskan perilakunya.

# Menguji pop saat Stack sudah dalam keadaan kosong
try:
    s.pop()
except IndexError as e:
    print("Terjadi error:", e)
```
