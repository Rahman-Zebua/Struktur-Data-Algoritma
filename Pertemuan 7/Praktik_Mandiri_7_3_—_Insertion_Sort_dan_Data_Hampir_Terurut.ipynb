```python
def insertion_sort(arr):
    data = arr.copy()
    n = len(data)

    comparisons = 0
    shifts = 0

    for i in range(1, n):
        key = data[i]
        j = i - 1

        while j >= 0:
            comparisons += 1
            if data[j] > key:
                data[j + 1] = data[j]  # Pergeseran data[j+1] = data[j]
                shifts += 1
                j -= 1
            else:
                break

        data[j + 1] = key

    return data, comparisons, shifts

# Data Uji
hampir = [1, 2, 3, 5, 4, 6, 7, 8]
acak = [7, 2, 8, 1, 5, 3, 6, 4]

# Eksekusi
hasil_hampir, comp_hampir, shift_hampir = insertion_sort(hampir)
hasil_acak, comp_acak, shift_acak = insertion_sort(acak)

# Output Data Hampir Terurut
print(f"Data Awal            : {hampir}")
print(f"Hampir Terurut       : {hasil_hampir}")
print(f"Comparison           : {comp_hampir}")
print(f"Shifts               : {shift_hampir}")

print()  # Jarak antar output

# Output Data Acak (Label disesuaikan)
print(f"Data Awal            : {hasil_acak}")
print(f"Hasil Acak           : {acak}")
print(f"Comparison           : {comp_acak}")
print(f"Shifts               : {shift_acak}")
```
