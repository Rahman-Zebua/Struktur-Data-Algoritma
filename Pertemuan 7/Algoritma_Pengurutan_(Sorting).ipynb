```python
#7.5. Bubble Sort
 #Bubble Sort membandingkan dua elemen yang berdekatan. Jika urutannya salah, keduanya
 #ditukar. Setelah satu pass penuh untuk urutan menaik, elemen terbesar pada bagian yang
 #belum terurut akan berpindah ke posisi paling kanan.
#7.5.2 Implementasi Python
def bubble_sort(data):
    n = len(data)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if data[j] > data[j + 1]:
                # Tukar posisi elemen jika elemen kiri lebih besar dari elemen kanan
                data[j], data[j + 1] = data[j + 1], data[j]
    return data

# Data uji coba
nilai = [5, 1, 4, 2, 8]
print(bubble_sort(nilai))
```

```python
#7.8. Selection Sort
 #Selection Sort membagi data menjadi bagian yang sudah terurut dan belum terurut. Pada
 #setiap iterasi, algoritma mencari nilai minimum dari bagian yang belum terurut, lalu
 #menukarnya dengan elemen pada posisi awal bagian tersebut.
def selection_sort(data):
    n = len(data)
    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            if data[j] < data[min_index]:
                min_index = j
        # Tukar elemen terkecil yang ditemukan dengan elemen di posisi i
        data[i], data[min_index] = data[min_index], data[i]
    return data

# Data uji coba
nilai = [29, 10, 14, 37, 13]
print(selection_sort(nilai))
```

```python
#7.9. Insertion Sort
 #Insertion Sort bekerja seperti menyusun kartu di tangan. Bagian kiri dianggap sudah terurut.
 #Elemen berikutnya diambil sebagai key lalu elemen yang lebih besar digeser ke kanan hingga
 #posisi yang tepat untuk key ditemukan.
def insertion_sort(data):
    for i in range(1, len(data)):
        key = data[i]
        j = i - 1
        # Geser elemen yang lebih besar dari key ke posisi kanan
        while j >= 0 and data[j] > key:
            data[j + 1] = data[j]
            j -= 1
        # Tempatkan key pada posisi yang tepat
        data[j + 1] = key
    return data

# Data uji coba
nilai = [5, 8, 12, 4, 9]
print(insertion_sort(nilai))
```
