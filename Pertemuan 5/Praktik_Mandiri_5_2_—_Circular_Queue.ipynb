```python
#2. Enqueue A, B, C, D dan cek kondisi penuh.

class CircularQueue:
    def __init__(self, kapasitas=4):
        self.kapasitas = kapasitas
        self.data = [None] * kapasitas
        self.front = 0
        self.rear = -1
        self.size = 0

    def enqueue(self, item):
        if self.size == self.kapasitas:
            print("Queue penuh")
            return

        self.rear = (self.rear + 1) % self.kapasitas
        self.data[self.rear] = item
        self.size += 1

    def is_full(self):
        return self.size == self.kapasitas

    def display(self):
        print(self.data)


queue = CircularQueue(4)

queue.enqueue("A")
queue.enqueue("B")
queue.enqueue("C")
queue.enqueue("D")

queue.display()
print("Queue penuh:", queue.is_full())
```

```python
#4. Dequeue dua elemen.

class CircularQueue:
    def __init__(self, kapasitas=4):
        self.kapasitas = kapasitas
        self.data = [None] * kapasitas
        self.front = 0
        self.rear = -1
        self.size = 0

    def enqueue(self, item):
        if self.size == self.kapasitas:
            print("Queue penuh")
            return

        self.rear = (self.rear + 1) % self.kapasitas
        self.data[self.rear] = item
        self.size += 1

    def dequeue(self):
        if self.size == 0:
            print("Queue kosong")
            return None

        item = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % self.kapasitas
        self.size -= 1
        return item

    def is_full(self):
        return self.size == self.kapasitas

    def display(self):
        print(self.data)


queue = CircularQueue(4)

queue.enqueue("A")
queue.enqueue("B")
queue.enqueue("C")
queue.enqueue("D")

print("Queue awal:")
queue.display()
print("Queue penuh:", queue.is_full())

queue.enqueue("E")

print("Dequeue pertama:", queue.dequeue())
print("Dequeue kedua:", queue.dequeue())

print("Queue setelah dequeue:")
queue.display()
```

```python
#6. Tampilkan items, front, rear, dan count setelah setiap operasi.
class CircularQueue:
    def __init__(self, kapasitas=4):
        self.kapasitas = kapasitas
        self.data = [None] * kapasitas
        self.front = 0
        self.rear = -1
        self.count = 0

    def enqueue(self, item):
        if self.count == self.kapasitas:
            print("Enqueue", item, ": Queue penuh")
            self.tampilkan()
            return

        self.rear = (self.rear + 1) % self.kapasitas
        self.data[self.rear] = item
        self.count += 1
        print("Enqueue", item)
        self.tampilkan()

    def dequeue(self):
        if self.count == 0:
            print("Dequeue: Queue kosong")
            self.tampilkan()
            return None

        item = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % self.kapasitas
        self.count -= 1
        print("Dequeue:", item)
        self.tampilkan()
        return item

    def tampilkan(self):
        print("Items:", self.data)
        print("Front:", self.front)
        print("Rear:", self.rear)
        print("Count:", self.count)
        print("--------------------")


queue = CircularQueue(4)

queue.enqueue("A")
queue.enqueue("B")
queue.enqueue("C")
queue.enqueue("D")
queue.enqueue("E")

queue.dequeue()
queue.dequeue()

queue.enqueue("E")
queue.enqueue("F")
```
