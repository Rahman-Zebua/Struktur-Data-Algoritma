```python
#5.6. Implementasi Awal Queue Menggunakan Python list
class SimpleQueue:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def enqueue(self, data):
        self.items.append(data)

    def dequeue(self):
        if self.is_empty():
            return None
        return self.items.pop(0)

    def peek(self):
        if self.is_empty():
            return None
        return self.items[0]

    def size(self):
        return len(self.items)
```

```python
#5.7. Menguji Queue Berbasis list
q = SimpleQueue()
print(q.is_empty())
q.enqueue('A')
q.enqueue('B')
q.enqueue('C')
print('Front:', q.peek())
print('Ukuran:', q.size())
print('Keluar:', q.dequeue())
print('Front baru:', q.peek())
print('Isi:', q.items)
```

```python
#5.10. Membangun Class Queue dengan deque
from collections import deque

class Queue:
    def __init__(self):
        self.items = deque()

    def is_empty(self):
        return len(self.items) == 0

    def enqueue(self, data):
        self.items.append(data)

    def dequeue(self):
        if self.is_empty():
            return None
        return self.items.popleft()

    def peek(self):
        if self.is_empty():
            return None
        return self.items[0]

    def size(self):
        return len(self.items)
```

```python
#5.11. Pengujian Operasi dan Edge Case
q = Queue()
for x in [10, 20, 30, 40]:
  q.enqueue(x)
print('Front:', q.peek())
while not q.is_empty():
  print('Diproses:', q.dequeue())
print('Dequeue saat kosong:', q.dequeue())
```

```python
#5.13. Queue Menggunakan Linked List
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedQueue:
    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0

    def is_empty(self):
        return self.front is None

    def enqueue(self, data):
        new_node = Node(data)
        if self.is_empty():
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node
        self.count += 1

    def dequeue(self):
        if self.is_empty():
            return None
        data = self.front.data
        self.front = self.front.next
        self.count -= 1
        if self.front is None:
            self.rear = None
        return data

    def peek(self):
        return None if self.is_empty() else self.front.data

    def size(self):
        return self.count
```

```python
#5.16. Implementasi Circular Queue Berkapasitas Tetap
class CircularQueue:
    def __init__(self, capacity):
        self.capacity = capacity
        self.items = [None] * capacity
        self.front = 0
        self.rear = 0
        self.count = 0

    def is_empty(self):
        return self.count == 0

    def is_full(self):
        return self.count == self.capacity

    def enqueue(self, data):
        if self.is_full():
            return False
        self.items[self.rear] = data
        self.rear = (self.rear + 1) % self.capacity
        self.count += 1
        return True

    def dequeue(self):
        if self.is_empty():
            return None
        data = self.items[self.front]
        self.items[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.count -= 1
        return data

    def peek(self):
        if self.is_empty():
            return None
        return self.items[self.front]

    def size(self):
        return self.count
```

```python
#5.17. Dry Run Circular Queue
cq = CircularQueue(5)

for x in ['A', 'B', 'C']:
  cq.enqueue(x)

print(cq.items, cq.front, cq.rear)
print('Keluar:', cq.dequeue())

cq.enqueue('D')
cq.enqueue('E')
cq.enqueue('F')

print(cq.items, cq.front, cq.rear)
print('Front:', cq.peek())
```

```python
#5.20. Studi Kecil Deque: Pemeriksaan Palindrome
from collections import deque


def is_palindrome(text):
    chars = deque(c.lower() for c in text if c.isalnum())
    while len(chars) > 1:
        if chars.popleft() != chars.pop():
            return False
    return True


print(is_palindrome("Katak"))
print(is_palindrome("Struktur Data"))
```

```python
#5.22. Mini Studi Kasus — Sistem Antrean Layanan
from collections import deque


class ServiceQueue:
    def __init__(self):
        self.queue = deque()
        self.next_number = 1

    def arrive(self, name):
        ticket = f"A{self.next_number:03d}"
        self.next_number += 1
        self.queue.append((ticket, name))
        return ticket

    def serve(self):
        if not self.queue:
            return None
        return self.queue.popleft()

    def waiting_count(self):
        return len(self.queue)

    def next_customer(self):
        return None if not self.queue else self.queue[0]
#5.22.1 Menjalankan Skenario
service = ServiceQueue()
print(service.arrive('Ani'))
print(service.arrive('Budi'))
print(service.arrive('Citra'))
print('Berikutnya:', service.next_customer())
print('Dilayani :', service.serve())
print('Menunggu :', service.waiting_count())
```
