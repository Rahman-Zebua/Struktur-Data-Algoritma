```python
#1. Buat class Queue berbasis collections.deque
from collections import deque

class Queue:
    def __init__(self):
        self.data = deque()

    def enqueue(self, item):
        self.data.append(item)

    def dequeue(self):
        if self.is_empty():
            return "Queue kosong"
        return self.data.popleft()

    def peek(self):
        if self.is_empty():
            return "Queue kosong"
        return self.data[0]

    def is_empty(self):
        return len(self.data) == 0

    def size(self):
        return len(self.data)

    def display(self):
        print(list(self.data))


queue = Queue()
queue.enqueue("A")
queue.enqueue("B")
queue.enqueue("C")

queue.display()
print("Data yang keluar:", queue.dequeue())
print("Data paling depan:", queue.peek())
print("Jumlah data:", queue.size())
```

```python
#6. Keluarkan seluruh elemen dengan loop sampai Queue kosong.
from collections import deque

class Queue:
    def __init__(self):
        self.data = deque()

    def enqueue(self, item):
        self.data.append(item)

    def dequeue(self):
        if self.is_empty():
            return None
        return self.data.popleft()

    def front(self):
        if self.is_empty():
            return None
        return self.data[0]

    def size(self):
        return len(self.data)

    def is_empty(self):
        return len(self.data) == 0

    def display(self):
        print(list(self.data))


queue = Queue()

queue.enqueue(10)
queue.enqueue(20)
queue.enqueue(30)
queue.enqueue(40)
queue.enqueue(50)

print("Front:", queue.front())
print("Size:", queue.size())

print("Dequeue pertama:", queue.dequeue())
print("Dequeue kedua:", queue.dequeue())

queue.enqueue(60)
queue.enqueue(70)

print("Queue sekarang:")
queue.display()

while not queue.is_empty():
    print("Data keluar:", queue.dequeue())

print("Queue kosong:", queue.is_empty())
```

```python
#7. Uji dequeue dan peek ketika Queue kosong.
from collections import deque

class Queue:
    def __init__(self):
        self.data = deque()

    def enqueue(self, item):
        self.data.append(item)

    def dequeue(self):
        if self.is_empty():
            return "Queue kosong"
        return self.data.popleft()

    def peek(self):
        if self.is_empty():
            return "Queue kosong"
        return self.data[0]

    def is_empty(self):
        return len(self.data) == 0

queue = Queue()

print("Dequeue:", queue.dequeue())
print("Peek:", queue.peek())
```
