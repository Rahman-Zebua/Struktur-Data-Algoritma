```python
#1. Implementasi fungsi is_palindrome()
from collections import deque

def is_palindrome(teks):
    teks = ''.join(huruf.lower() for huruf in teks if huruf.isalnum())
    dq = deque(teks)

    while len(dq) > 1:
        kiri = dq.popleft()
        kanan = dq.pop()

        if kiri != kanan:
            return False

    return True
```

```python
#2. Menguji kata level, katak, radar, python, dan algoritma
kata = ["level", "katak", "radar", "python", "algoritma"]

for teks in kata:
    print(teks, ":", is_palindrome(teks))
```

```python
#3. Menguji kalimat yang mengandung spasi atau tanda baca setelah normalisasi.
kalimat = [
    "Kasur ini rusak",
    "A man, a plan, a canal: Panama!",
    "Ibu Ratna antar ubi"
]

for teks in kalimat:
    print(teks, ":", is_palindrome(teks))
```

```python
#5. Mengubah fungsi agar menampilkan pasangan kiri-kanan pada setiap iterasi.
from collections import deque

def is_palindrome(teks):
    teks = ''.join(huruf.lower() for huruf in teks if huruf.isalnum())
    dq = deque(teks)

    while len(dq) > 1:
        kiri = dq.popleft()
        kanan = dq.pop()

        print("Kiri:", kiri, "| Kanan:", kanan)

        if kiri != kanan:
            print("Bukan palindrome")
            return False

    print("Palindrome")
    return True

print(is_palindrome("katak"))
```

```python
#6. Mengapa Deque lebih natural daripada Queue satu arah untuk masalah palindrome?
Deque lebih cocok digunakan untuk memeriksa palindrome karena Deque memungkinkan pengambilan karakter dari kedua ujung, yaitu depan dan belakang. Dengan demikian, karakter paling kiri dan paling kanan dapat dibandingkan secara langsung. Sementara itu, Queue satu arah hanya mengambil data dari bagian depan sehingga kurang praktis untuk membandingkan karakter dari kedua ujung.
```

```python
#7. Kompleksitas waktu dan ruang fungsi is_palindrome().
Kompleksitas waktu: O(n), karena fungsi memeriksa pasangan karakter dari kedua ujung hingga mencapai bagian tengah kata atau kalimat.

Kompleksitas ruang: O(n), karena karakter yang sudah dinormalisasi disimpan dalam Deque dengan jumlah elemen sesuai panjang teks.

Keterangan: n adalah jumlah karakter dalam teks setelah normalisasi.
```
